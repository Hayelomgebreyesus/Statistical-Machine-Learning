```python
mat= np.array([[1,2,3,],[3,5,6,],[7,8,9,]])
print(f'the maximum value in this matrix is :{mat.max()}')
print(f'the sum of the first row is: {np.sum(mat[0,:])}')
print(f'the maximum element of the first row is: {np.max(mat[0,:])}')
matt= np.transpose(mat)
print(matt)
inve=np.linalg.inv(mat)
print(inve)
```
