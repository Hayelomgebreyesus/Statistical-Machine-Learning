```python
mat= np.array([[1,2,3,4],[4,5,6,7],[7,8,9,8]])
print(f'the first element in the second row is: {mat[1,0]}')
print(f'all elements of the third row are : {mat[2,:]}')
mat1=np.array([3,2,1])
#mat[0,:]=mat1
#print(mat)
re_mat=np.reshape(mat,(6,2))
print(f'the new reshaped matrix is: {re_mat} ')
z= np.zeros(2)
x=np.reshape(z,(2,1))
row_added=np.vstack((re_mat,z))
#col_added=np.hstack((re_mat,x))
print(f'the new matrix is :{row_added}')
```

```python
mat= np.array([[1,2,3,],[3,5,6,],[7,8,9,]])
print(f'the maximum value in this matrix is :{mat.max()}')
print(f'the sum of the first row is: {np.sum(mat[0,:])}')
print(f'the maximum element of the first row is: {np.max(mat[0,:])}')
matt= np.transpose(mat)
print(matt)
inve=np.linalg.inv(mat)
print(inve)
```

```python
num_students_passed=np.array([20,30,40,25,32,60,50,80,70,70])
print(np.shape(num_students_passed))
year=np.array([2010,2011,2012,2013,2014,2015,2016,2017,2018,2019])
print(np.shape(year))
plt.bar(year,num_students_passed)
plt.title("number of students passed from 2011-2018")
plt.xlim(2010, 2025)

plt.xlabel("year")
plt.ylabel("number students passed")
plt.show()
```
